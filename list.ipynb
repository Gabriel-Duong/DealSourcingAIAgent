```python
import requests
```

```python
BASE_URL = "https://expara.decilehub.com/api/v1"
TOKEN = "4ZW8M6LEvTQ8peVqFU9StLdo"

headers = {
    "Authorization": f"Bearer {TOKEN}",
    "Content-Type": "application/json"
}
```

```python
endpoints = [
    "/events",
    "/files",
    "/folders",
    "/organizations",
    "/people",
    "/pipeline_prospects",
    "/pipelines"
]
```

```python
resp = requests.get(f"{BASE_URL}/organizations", headers=headers)
print(resp.json())
```

```python
for ep in endpoints:
    url = f"{BASE_URL}{ep}"
    r = requests.get(url, headers=headers)
    print(f"Endpoint {ep}: status={r.status_code}")
    try:
        print(r.json())
    except Exception:
        print(r.text)
```
